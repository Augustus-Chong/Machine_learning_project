```python
import torch

# Load model
model = torch.hub.load('ultralytics/yolov5', 'yolov5s', pretrained=True)

# --- NEW: Lower the confidence threshold (0.25 is default, try 0.1) ---
model.conf = 0.1 

# Run Inference
results = model('fpc.jpg')

# Show results
results.show()
```
